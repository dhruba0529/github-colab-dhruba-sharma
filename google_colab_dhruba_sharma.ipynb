```python
import pandas as pd

data = pd.read_csv("robot_mission.csv")
numbers = data["number"].tolist()
rounds = data["round"].tolist()
print(numbers)
print(rounds)
```

```python
total = 0
positions = []
for i in range(len(numbers)):
  if numbers[i] % 2 == 0:
    total += numbers[i]

  else:
    total -= numbers[i]
  positions.append(total)
  print(numbers[i],total)
```

```python
import matplotlib.pyplot as plt
plt.plot(rounds,positions)
plt.xlabel("rounds")
plt.ylabel("positions")
plt.axhline()
plt.title("Robots Movement")
plt.show()
```
